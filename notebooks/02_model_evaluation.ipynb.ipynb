Now the Model being trained, lets evaluate it

Loading the libraries and model trained.

```python
import joblib
from sklearn.metrics import r2_score, mean_absolute_error

model = joblib.load('Trained_HH_Model.joblib')
X_test, y_test = joblib.load('Test_Dataset.joblib')

print("Model and test data loaded successfully!")
```

Evaluating the model

```python
# predictions on the test data
predictions = model.predict(X_test)

# Calculate the evaluation metrics
r2 = r2_score(y_test, predictions)
mae = mean_absolute_error(y_test, predictions)

print(f"--- Model Performance ---")
print(f"R-squared (R²): {r2:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}\n")

print(f"The model can explain {r2:.1%} of the variation in the habitability score.")
print(f"On average, the model's predictions are off by {mae:.2f} points.")
```
